# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

```python
# Saisie des données par l'utilisateur
prix_ht = float(input("Entrez le prix unitaire HT : "))
taux_tva = float(input("Entrez le taux de TVA (ex: 20 pour 20%) : "))
nombre_articles = int(input("Entrez le nombre d'articles : "))

# Calcul du montant TTC
montant_ht = prix_ht * nombre_articles
montant_ttc = montant_ht * (1 + taux_tva / 100)

# Affichage du résultat
print(f"Le montant total TTC est de : {montant_ttc:.2f} €")
```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
# Correction : la cellule contenait le code du calcul de TVA (copier-coller de l'exercice 1)

# Saisie de la durée en secondes
duree = int(input("Entrez une durée en secondes : "))

heures = duree // 3600              # division entière : nombre d'heures complètes (1 h = 3600 s)
minutes = (duree % 3600) // 60      # reste après les heures, converti en minutes
secondes = duree % 60               # reste après les minutes

# Exemple : 12334 -> 3 heures, 25 minutes et 34 secondes
print(f"{duree} secondes = {heures} heures, {minutes} minutes et {secondes} secondes")
```

# 3. Conversion de durées (2)

Ecrivez un programme demandant une durée en heures, minutes, secondes à l’utilisateur et qui la convertit en secondes.<br>
Affichez le résultat.<br>
Par exemple 3 heures, 25 minutes et 34 secondes deviendra 12334 secondes.

```python
# Correction : la cellule contenait le code du calcul de TVA (copier-coller de l'exercice 1)

# Saisie de la durée en heures, minutes et secondes
heures = int(input("Nombre d'heures : "))
minutes = int(input("Nombre de minutes : "))
secondes = int(input("Nombre de secondes : "))

# Conversion : 1 h = 3600 s et 1 min = 60 s
total_secondes = heures * 3600 + minutes * 60 + secondes

# Exemple : 3 h 25 min 34 s -> 12334 secondes
print(f"{heures} heures, {minutes} minutes et {secondes} secondes = {total_secondes} secondes")
```
