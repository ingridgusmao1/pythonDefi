#### 1.2 Maintenant, vous allez devoir convertir des objets d'un certain type, en un autre type d'objet.

Convertissez la variable `count` en float.

```python
count = 5
count_float = float(count)   # conversion int -> float
print(count_float, type(count_float))   # 5.0 <class 'float'>
```

Convertissez la variable `price` en int

```python
price = 5.4
price_int = int(price)   # conversion float -> int
print(price_int, type(price_int))   # 5 <class 'int'>
```

Que pouvez-vous remarquer ?

Réponse :
`int()` **n'arrondit pas**, il **tronque** : il supprime simplement la partie décimale.
`int(5.4)` donne `5`, mais `int(5.9)` donnerait aussi `5` (et non 6). On perd donc de l'information lors de la conversion.

Trouvez le moyen d'arrondir la variable `temperature`.

```python
temperature = 5.3

# round() arrondit à l'entier le plus proche
temperature_arrondie = round(temperature)
print(temperature_arrondie)          # 5
print(type(temperature_arrondie))    # <class 'int'>
```

Après avoir arrondi la variable `temperature`, est-ce que c'est toujours un `float` ?

Réponse :
Non. `round(5.3)` (sans deuxième argument) renvoie un **`int`** : `5`.
Remarque : si on précise un nombre de décimales, par exemple `round(5.3, 0)` ou `round(5.37, 1)`, le résultat reste un `float` (`5.0`, `5.4`).

Ci-dessous, est-ce que height est un `int` ?

```python
height = 6.0
print(type(height))          # <class 'float'>
print(height.is_integer())   # True : la valeur est entière... mais le type reste float
```

Réponse :
Non, `height` est un **`float`** : c'est la présence du point décimal (`6.0`) qui détermine le type, pas la valeur.
Même si sa valeur est « ronde », Python le stocke comme un nombre décimal. Pour obtenir un `int`, il faudrait écrire `int(height)`.

#### 3.2 Inversion de deux variables

Réalisez les étapes suivantes dans l'ordre :
 - Demande à l'utilisateur de rentrer deux valeurs et stocke-les dans deux variables différentes (par exemple `x` et `y`).
 - Effectue l'échange des valeurs entre ces deux variables.
 - Affiche la valeur des deux variables.

```python
# Demander deux valeurs à l'utilisateur
x = input("Entrez la première valeur : ")
y = input("Entrez la deuxième valeur : ")

# Échanger les valeurs grâce à une variable temporaire
temp = x   # on sauvegarde x avant de l'écraser
x = y
y = temp

# Remarque : en Python, on peut aussi faire l'échange en une ligne : x, y = y, x

# Afficher les résultats
print("Après inversion :")
print("x =", x)
print("y =", y)
```
