# Mission 2

Create a "fruit_vendors" dataframe that reflects the below example **using a dictionary** data structure

![](https://i.imgur.com/CHPn7ZF.png)

[This resource could help you.](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.from_dict.html)

```python
# ⚠️ Les valeurs ci-dessous sont à adapter au tableau de l'image de l'énoncé
# Avec un dictionnaire : chaque clé devient une COLONNE, chaque liste les valeurs de cette colonne
donnees_vendeurs = {
    "Apples": [35, 41],
    "Bananas": [21, 34],
}
fruit_vendors = pd.DataFrame(donnees_vendeurs, index=["2017 Sales", "2018 Sales"])
# Variante avec la méthode citée dans l'énoncé : pd.DataFrame.from_dict(donnees_vendeurs)
fruit_vendors
```

# Mission 3

Use a loop to display the values from each column in the dataframe from mission 2.

[This resource could help you](https://www.w3schools.com/python/pandas/ref_df_items.asp)

```python
# items() renvoie, pour chaque colonne, son nom et la Series de ses valeurs
for nom_colonne, valeurs in fruit_vendors.items():
    print(f"Colonne : {nom_colonne}")
    for index, valeur in valeurs.items():
        print(f"   {index} -> {valeur}")
```

# Mission 4

Delete the first row of the dataframe from mission 2
