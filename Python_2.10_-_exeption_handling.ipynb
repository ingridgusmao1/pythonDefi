You must correct each of the scripts below **using exception handling** and **without modifying lists**.

Script number 1

```python
list_integers = [1, 12, -45, 6, 27, 0, 24]
for i in list_integers:
  try:
    print(1/i)
  except ZeroDivisionError:
    # La division par 0 est impossible : on affiche un message et la boucle continue
    print("Erreur : division par zéro impossible pour", i)
```

Script number 2

```python
list_integers = [1, 12, -45, 6, 27, 24]
for i in range(10):
  try:
    print(1/list_integers[i])
  except IndexError:
    # La liste ne contient que 6 éléments (index 0 à 5) : les index 6 à 9 n'existent pas
    print(f"Erreur : l'index {i} n'existe pas dans la liste")
```

Script number 3

```python
list_integers = [1, 12, -45, 6, 27, 0, 24]
for i in range(10):
  try:
    print(1/list_integers[i])
  except ZeroDivisionError:
    # Deux erreurs différentes peuvent survenir : on les gère chacune avec son propre except
    print(f"Erreur : division par zéro (index {i})")
  except IndexError:
    print(f"Erreur : l'index {i} n'existe pas dans la liste")
```

Script number 4

```python
list_words = ['hello', 'good morning', 'GOOD Evening', 34, 'Bye' ]

for word in list_words:
  try:
    print(word.upper())
  except AttributeError:
    # 34 est un int : il n'a pas de méthode upper() (réservée aux str)
    print(f"Erreur : {word} n'est pas une chaîne de caractères ({type(word).__name__})")
```
