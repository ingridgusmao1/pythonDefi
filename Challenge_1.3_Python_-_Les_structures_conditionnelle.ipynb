# 1. Calcul du montant d'une remise

Un commerçant accorde une remise de 5 % pour tout achat d’un montant compris entre 100 et 500 € et 8 % au-delà.<br>
Ecrivez un programme de calcul du montant de la remise sur un achat donné.<br>
Affichez le montant de la remise.

```python
# Saisie du montant de l'achat
montant = float(input("Entrez le montant de l'achat : "))

# Calcul de la remise
if montant < 100:
    remise = 0.0
elif montant <= 500:
    remise = montant * 0.05
else:
    remise = montant * 0.08

# Affichage du montant de la remise
print(f"Le montant de la remise est de : {remise:.2f} €")
```

# 2. Savoir si trois entiers sont triés

Ecrivez un programme faisant saisir trois entiers x, y, z à l’utilisateur, et lui indiquer si ces nombres sont dans l’ordre croissant (x <= y <= z).

```python
# Correction : la cellule contenait le code de la remise (copier-coller de l'exercice 1)

# Saisie des trois entiers
x = int(input("Entrez x : "))
y = int(input("Entrez y : "))
z = int(input("Entrez z : "))

# Python permet d'enchaîner les comparaisons : x <= y <= z équivaut à (x <= y) and (y <= z)
if x <= y <= z:
    print("Les nombres sont dans l'ordre croissant.")
else:
    print("Les nombres ne sont pas dans l'ordre croissant.")
```

# 3. Tri de trois réels

Ecrivez un programme faisant saisir trois nombres réels x, y, z à l’utilisateur et qui les trie par ordre croissant (à la fin du déroulement du programme x ≤ y ≤ z).<br>
Affichez x, y et z.

```python
# Correction : la cellule contenait le code de la remise (copier-coller de l'exercice 1)

# Saisie des trois réels
x = float(input("Entrez x : "))
y = float(input("Entrez y : "))
z = float(input("Entrez z : "))

# Tri par échanges successifs (uniquement avec des conditions) :
# 1) on s'assure que x <= y
if x > y:
    x, y = y, x
# 2) on s'assure que y <= z  -> z contient maintenant le plus grand
if y > z:
    y, z = z, y
# 3) le nouveau y peut être plus petit que x : on revérifie x <= y
if x > y:
    x, y = y, x

print("x =", x)
print("y =", y)
print("z =", z)
```

# 4. Signe d'un produit

Ecrivez un programme qui affiche le signe du produit de deux nombres réels sans calculer la valeur de ce produit. Par signe, on entend positif, négatif ou nul.

```python
# Saisie des deux nombres réels
x = float(input("Entrez le premier nombre : "))
y = float(input("Entrez le deuxième nombre : "))

# Vérification si l'un des nombres est égal à zéro
if x == 0 or y == 0:
    print("Le produit est nul.")
# Vérification si les deux nombres ont le même signe (tous deux positifs ou tous deux négatifs)
elif (x > 0 and y > 0) or (x < 0 and y < 0):
    print("Le produit est positif.")
# Sinon, l'un est positif et l'autre est négatif
else:
    print("Le produit est négatif.")
```
