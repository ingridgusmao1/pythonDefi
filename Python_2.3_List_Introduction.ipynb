```python
names = ["Fredy", "Nicolas", "Alexandre"]

for i in enumerate(names):
  print(i)
```

1. What type of object is returned as output?  
Hint: You can use `type(i)`.

Answer:
Un **`tuple`** (`<class 'tuple'>`). À chaque tour de boucle, `enumerate()` renvoie un tuple de 2 éléments : `(index, valeur)`, par exemple `(0, 'Fredy')`.

```python
for i, val in enumerate(names):
  print(val)
```

2. What does `i` correspond to?  
What does `val` correspond to?  
Why is it so?

Answer:
- `i` correspond à **l'index** (la position : 0, 1, 2).
- `val` correspond à **la valeur** de l'élément (le prénom).

C'est l'**affectation parallèle (unpacking)** : chaque tuple `(index, valeur)` renvoyé par `enumerate()` est « déballé » dans les deux variables, dans l'ordre. Le 1er élément du tuple va dans `i`, le 2e dans `val`.

```python
# Démonstration : on affiche les deux variables et le type de ce que renvoie enumerate
for i, val in enumerate(names):
  print(i, "->", val)

for element in enumerate(names):
  print(type(element))   # <class 'tuple'>
```

```python
for val, i in enumerate(names):
  print(val)
```

3. This time, `i` returns first names and `val` returns numbers. Why is it so?

Answer:
Les noms de variables n'ont **aucun sens pour Python** : seule la **position** compte lors du déballage.
Le tuple renvoyé par `enumerate()` est toujours `(index, valeur)`. La première variable écrite (ici `val`) reçoit donc l'index (un nombre) et la deuxième (ici `i`) reçoit le prénom. C'est pour cela qu'on choisit des noms explicites, dans le bon ordre.

4. What does the built in function `enumerate` do?
In what cases can it be useful?

Answer:
`enumerate()` est une fonction intégrée qui parcourt un itérable (liste, chaîne, tuple…) en ajoutant un **compteur** : elle renvoie des paires `(index, élément)`. On peut choisir le point de départ du compteur : `enumerate(names, 1)`.

Utile quand on a besoin **à la fois de la position et de la valeur** : afficher une liste numérotée, retrouver la position d'un élément, modifier un élément de la liste à un index donné, etc. Cela évite d'écrire `for i in range(len(liste))` puis `liste[i]`.

# Mission 2

```python
test_list = ['Hi', 'my', 'name', 'is', "hi", "my","name", "is", "chickychicky"]

test_list[8] = "Marshall"

test_list
```

1. What does the 8 in square brackets mean?  
What action has been performed?

Answer:
Le `8` est l'**index** de l'élément visé, c'est-à-dire le **9e élément** de la liste (l'indexation commence à 0) : `"chickychicky"`.
L'action effectuée est une **réaffectation (modification)** : on remplace la valeur à l'index 8 par `"Marshall"`. C'est possible car une liste est **mutable**.

```python
print(test_list[8])    # 'Marshall'
print(len(test_list))  # 9 éléments -> indices de 0 à 8
```

```python
for i in range(4):
  print(test_list[i])
```

2. What does `range(4)` mean?  
Why are only the first 4 elements of `test_list` displayed?

Answer:
`range(4)` génère la suite d'entiers **0, 1, 2, 3** : on part de 0 et on s'arrête **avant** 4 (la borne de fin est exclue).
La boucle affiche donc `test_list[0]` à `test_list[3]`, soit les 4 premiers éléments.

```python
print(list(range(4)))         # [0, 1, 2, 3]
print(list(range(2, 8, 2)))   # [2, 4, 6]
```

```python
for i in range(2, 8, 2):
  print(test_list[i])
```

3. What does `range(2,8,2)` mean ?  
Why are only the elements 'name', 'hi' and 'name' displayed?

Answer:
`range(2, 8, 2)` = `range(début, fin, pas)` : on commence à **2**, on s'arrête **avant 8**, en avançant de **2 en 2** → `2, 4, 6`.
On affiche donc `test_list[2]` = `'name'`, `test_list[4]` = `'hi'` et `test_list[6]` = `'name'`.

```python
numbers = [2,3,1,5,9]

for i in range(10,15):
  numbers.append(i)

numbers
```

1. What does `range(10,15)` mean ?  
  What is the purpose of the method `append()` ?

  

Answer:
- `range(10, 15)` génère les entiers de **10 à 14** (15 est exclu).
- `append()` est une méthode de liste qui **ajoute un élément à la fin** de la liste. Elle modifie la liste directement (« en place ») et ne renvoie rien (`None`).

```python
# Démonstration de del
numbers = [2, 3, 1, 5, 9, 10, 11, 12, 13, 14]

del numbers[0]       # supprime l'élément à l'index 0 (le 2)
print(numbers)

del numbers[-3:]     # supprime une tranche : les 3 derniers éléments
print(numbers)

copie = numbers
del copie            # supprime la variable "copie" elle-même (le nom n'existe plus)
```

2. What is the purpose of the `del` command?

Answer:
`del` est une instruction (un mot-clé, pas une méthode) qui **supprime** :
- un élément d'une liste à partir de son index : `del liste[0]`
- une tranche d'éléments : `del liste[2:5]`
- une clé d'un dictionnaire : `del dico["cle"]`
- ou une variable entière : `del ma_variable` (elle n'existe plus ensuite).

```python
import random

random.randint(0,10)
```

1. What can the method `randint()` do ?

Answer:
`randint(a, b)` (fonction du module `random`) renvoie un **entier aléatoire compris entre `a` et `b`, bornes incluses**. Ici, `random.randint(0, 10)` peut donc renvoyer n'importe quel entier de 0 à 10 (contrairement à `range()`, la borne de fin est incluse).

```python
# Démonstration de .insert()
fruits = ["pomme", "banane", "cerise"]
fruits.insert(1, "kiwi")   # arguments : 1 (index) et "kiwi" (élément)
print(fruits)              # ['pomme', 'kiwi', 'banane', 'cerise']
```

2. What can the method `.insert()` do ?  
You must answer using the terms 'parameter' and 'argument'.

Answer:
La méthode `.insert()` **ajoute un élément à une position précise** d'une liste ; les éléments suivants sont décalés vers la droite.

Elle possède **deux paramètres** : `index` (la position où insérer) et `object` (l'élément à insérer) → `liste.insert(index, object)`.
Quand on l'appelle, les valeurs qu'on lui transmet sont les **arguments** : dans `fruits.insert(1, "kiwi")`, `1` est l'argument passé au paramètre `index` et `"kiwi"` l'argument passé au paramètre `object`.

1. What can the `.join()` method do?  
What is the type of `test_list` before using `.join()`?  
What is the type of `test_list` after using `.join()`?

Answer:
- `.join()` est une méthode de **chaîne** : elle **assemble tous les éléments d'un itérable de chaînes en une seule chaîne**, en plaçant entre eux le séparateur sur lequel on l'appelle (ici `" "`, un espace).

- Après : `test_list` est une **`str`** (`'Hi my name is hi my name is chickychicky'`).

2. What can the `.split()` method do?  
What is the type of `christmas` before using `.split()`?  
What is the type of `christmas` after using `.split()`?

Answer:
- `.split()` **découpe une chaîne en une liste de sous-chaînes**. Sans argument, elle coupe sur les espaces (et retours à la ligne) ; on peut aussi préciser un séparateur.
- Avant : `christmas` est une **`str`**.
- Après : `christmas.split()` renvoie une **`list`** de mots. Attention : `christmas` elle-même reste une `str` (les chaînes sont immuables) ; seule la valeur renvoyée est une liste. Pour la garder, il faudrait écrire `christmas = christmas.split()`.

```python
christmas = "Noël c'est bien, mais attention à ne pas manger trop de chocolat. On le sait, mais on le fait à chaque fois. C'est la vie"


christmas.split("à")
```

3. Why does the `.split()` method split the `string` into 3 elements?

Answer:
Parce que le séparateur `"à"` apparaît **2 fois** dans la phrase (« attention **à** ne pas… » et « **à** chaque fois »). Chaque occurrence crée une coupure : 2 coupures → **3 morceaux**. Le séparateur lui-même disparaît du résultat.
