# Mission 2

Modify, using a `numpy` method, the `array_1` object to be an object that has two axes, including 1 row and 32 columns.  
Save it under the variable `array_2`.

```python
# reshape(lignes, colonnes) : 1 ligne et 32 colonnes
array_2 = array_1.reshape(1, 32)
print(array_2.shape)   # (1, 32)
array_2
```

Now, modify `array_1` into an object that has 32 rows and only 1 column.  
Save it under the variable `array_3`.

```python
# 32 lignes et 1 colonne
array_3 = array_1.reshape(32, 1)
print(array_3.shape)   # (32, 1)
array_3[:5]   # affichage des 5 premières lignes
```

Access the first element of `array_2`, then of `array_3`.  
What do you notice?  
Why ?

```python
print(array_2[0])   # toute la première LIGNE : un tableau de 32 valeurs
print(array_3[0])   # la première ligne de array_3 : un tableau d'1 seule valeur [10]

# Constat : dans les deux cas on n'obtient pas le nombre 10 seul, mais un tableau.
# Pourquoi : ce sont des tableaux à 2 axes ; un seul index sélectionne une LIGNE entière.
# array_2 n'a qu'une ligne (qui contient tout), array_3 a 32 lignes d'un seul élément.
# Pour obtenir la valeur seule il faut deux index (ligne, colonne) :
print(array_2[0, 0], array_3[0, 0])
```

# Mission 3

Access the value that is at index 3 of array `array_2`.

# Mission 4

Reshape `array_1` to be a 4-row, 8-column array,
rather than a 32-element vector.  
Save your array under the variable `array_4`.

```python
array_4 = array_1.reshape(4, 8)
array_4
```

Access for each row, the values ​​of the columns which have the index 1 to 2 included.

```python
# Toutes les lignes (:), colonnes d'index 1 à 2 inclus -> 1:3 (la borne de fin est exclue)
array_4[:, 1:3]
```

Access for rows at position 1 and 2 inclusive, the values ​​of all columns.

```python
# "Position" 1 et 2 = 1re et 2e lignes, soit les index 0 et 1 (les positions commencent à 1, les index à 0)
array_4[0:2, :]
# Si on interprète "position" comme un index : array_4[1:3, :]
```

Access for the rows which have the index 2 to 3 inclusive, to the values ​​of the columns which have the index 3 to 5 inclusive.

```python
# Lignes d'index 2 à 3 inclus -> 2:4 ; colonnes d'index 3 à 5 inclus -> 3:6
array_4[2:4, 3:6]
```

Select the first column, as well as the columns whose index is a multiple of 2.

```python
# Pas de 2 à partir de 0 : colonnes 0, 2, 4, 6 (la première colonne, index 0, est bien un multiple de 2)
array_4[:, ::2]
```

Select columns whose index is odd.

```python
# On commence à l'index 1, avec un pas de 2 : colonnes 1, 3, 5, 7
array_4[:, 1::2]
```
