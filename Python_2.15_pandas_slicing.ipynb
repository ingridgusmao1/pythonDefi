# Mission 1

Display the following information [for this dataset](https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/u.user).
- The last 5 rows
- The first 5 rows
- The type of each column

Note : start coding something by typing `pd.read_csv()` and explore the different options available, especially the `sep` parameter.


```python
import pandas as pd

# Le fichier n'a pas d'en-tête et les colonnes sont séparées par "|" -> sep="|", header=None
users = pd.read_csv("https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/u.user",
                    sep="|", header=None,
                    names=["user_id", "age", "gender", "occupation", "zip_code"])

print("5 dernières lignes :")
display(users.tail())
print("5 premières lignes :")
display(users.head())
print("Type de chaque colonne :")
print(users.dtypes)
```

# Mission 5

Rename the columns [of this dataset](https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/ufo.csv), try using the following 2 different methods to achieve this:
- Setting column names directly when reading the file
- Editing after creating the dataframe ([This resource could be useful](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.rename.html))

```python
url_ufo = "https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/ufo.csv"
nouveaux_noms = ["city", "colors_reported", "shape_reported", "state", "time"]

# Méthode 1 : noms définis à la lecture du fichier
# header=0 indique que la 1re ligne contient les anciens noms, qui sont remplacés par names
ufo_1 = pd.read_csv(url_ufo, names=nouveaux_noms, header=0)
display(ufo_1.head(3))

# Méthode 2 : renommer après création avec rename() et un dictionnaire {ancien: nouveau}
ufo = pd.read_csv(url_ufo)
ufo = ufo.rename(columns={"City": "city",
                          "Colors Reported": "colors_reported",
                          "Shape Reported": "shape_reported",
                          "State": "state",
                          "Time": "time"})
display(ufo.head(3))
```

# Mission 6

Select the first 3 rows and all the columns of the dataframe using iloc

```python
# iloc : sélection par POSITION, la borne de fin est exclue -> lignes 0, 1, 2
ufo.iloc[0:3, :]
```

# Mission 7

Achieve the same result as in mission 6 by using loc

```python
# loc : sélection par LABEL, la borne de fin est INCLUSE -> labels 0, 1, 2
ufo.loc[0:2, :]
```
