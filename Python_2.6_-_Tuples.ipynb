# Mission 1

Create a tuple of your choice containing 10 elements in this order:
- 3 integers
- 3 strings
- 2 lists containing 3 items each
- 2 dictionaries containing 3 items (3 key/value pairs) each


```python
mon_tuple = (
    7, 21, 1984,                                         # 3 entiers
    "Paris", "Lyon", "Bordeaux",                         # 3 chaînes
    ["pomme", "poire", "kiwi"],                          # liste 1 (3 éléments)
    ["rouge", "vert", "bleu"],                           # liste 2 (3 éléments)
    {"nom": "Alice", "age": 30, "ville": "Nantes"},      # dictionnaire 1 (3 paires)
    {"langage": "Python", "version": 3, "type": "interprété"},  # dictionnaire 2 (3 paires)
)
print(mon_tuple)
print(len(mon_tuple))   # 10
```

Now, access the second-to-last item in the second list.

```python
# La 2e liste est à l'index 7 du tuple ; [-2] = avant-dernier élément de cette liste
print(mon_tuple[7][-2])   # 'vert'
```

# Mission 2

In the same tuple, access the last 4 elements using slicing.

```python
# Slicing avec un index négatif : du 4e élément en partant de la fin jusqu'à la fin
print(mon_tuple[-4:])
```

# Mission 3

Change the value of the 4th element (the first string) of your tuple (since tuples are immutable, this means recreating a tuple where this element is changed).

```python
# Un tuple est immuable : mon_tuple[3] = "Marseille" provoquerait une TypeError.
# On reconstruit donc un nouveau tuple en concaténant des morceaux :
mon_tuple = mon_tuple[:3] + ("Marseille",) + mon_tuple[4:]   # ("Marseille",) : la virgule crée un tuple d'1 élément
print(mon_tuple)

# Autre méthode : passer par une liste (mutable) puis revenir en tuple
# temp = list(mon_tuple); temp[3] = "Marseille"; mon_tuple = tuple(temp)
```

# Mission 4

Create 2 tuples of the same length that contain only integers.

Then create a script that compares the sum of their elements and displays the tuple that has the highest total.

```python
tuple_1 = (4, 8, 15, 16, 23)
tuple_2 = (42, 1, 3, 5, 7)

somme_1 = sum(tuple_1)
somme_2 = sum(tuple_2)
print("Somme tuple_1 :", somme_1, "| Somme tuple_2 :", somme_2)

if somme_1 > somme_2:
    print("Le tuple avec le plus grand total est :", tuple_1)
elif somme_2 > somme_1:
    print("Le tuple avec le plus grand total est :", tuple_2)
else:
    print("Les deux tuples ont le même total :", tuple_1, tuple_2)
```

# Mission 5

Here's a tuple:

`my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")`

Ask the user for a string, then for an integer position. Create an altered copy of `my_tuple` where the element at the given position is now the given string.

```python
my_tuple = ("data analyst", "data scientist", "data engineer", "data architect")

texte = input("Entrez une chaîne : ")
position = int(input("Entrez une position (index entre 0 et 3) : "))

# On convertit en liste (modifiable), on remplace, puis on revient en tuple
temp = list(my_tuple)
temp[position] = texte
nouveau_tuple = tuple(temp)

print(nouveau_tuple)
print(my_tuple)   # l'original n'a pas changé
```
