# Mission 1

Define a list, then display only the elements (and their position) that have even positions (0, 2, 4, 6...).

With `test_list = ["a", "b", "c", "d", "e"]`, the script should display:
```
a at position 0
c at position 2
e at position 4
```

````python
test_list = ["a", "b", "c", "d", "e"]

# range(0, len(test_list), 2) -> 0, 2, 4 : uniquement les positions paires
for position in range(0, len(test_list), 2):
    print(test_list[position], "at position", position)
````

### Mission 2  

Define a list, then **ask** for an index, followed by a string. Then, replace the list element at that **index** with the given string and display the resulting list.  

For example, with the list `test_list2 = ['hello', 'good morning', 'bye bye', 'have a good day']`, the index `2`, and the string `yeah`, you should display:

['hello', 'good morning', 'yeah', 'have a good day']

# Mission 3

Define a list with five numbers, then ask the user for a number. Put the number at the end of the list, remove the first element from the list, and finally display the list.

For example, with the list `test_list3 = [1, 9, 6, 15, 4]` and the input `7`, you should display:
```
[9, 6, 15, 4, 7]
```


````python
test_list3 = [1, 9, 6, 15, 4]

nombre = int(input("Entrez un nombre : "))

test_list3.append(nombre)   # ajout à la fin
test_list3.pop(0)           # suppression du premier élément (index 0)
print(test_list3)           # avec 7 : [9, 6, 15, 4, 7]
````

# Mission 4

Define a list, then ask the user for a string. Then, insert the string in the list at a random position, and display the resulting list.

For example, with the list `test_list4 = ["p", "y", "t", "h", "o"]` and the input `n`, you *might* get:
```
["p", "n", "y", "t", "h", "o"]
```
... but you might also get:
```
["p", "y", "t", "n", "h", "o"]
```


````python
import random

test_list4 = ["p", "y", "t", "h", "o"]

texte = input("Entrez une chaîne : ")

# Position aléatoire entre 0 et len(liste) inclus (len = insertion à la fin)
position = random.randint(0, len(test_list4))
test_list4.insert(position, texte)
print(test_list4)
````

# Mission 5

Create a list of integers. Then, display a list with the elements ordered from smallest to largest, using the right built-in function.

With the list `test_list5 = [3, 4, 0, -1, 35, 7]`, this will display:
```
[-1, 0, 3, 4, 7, 35]
```

````python
test_list5 = [3, 4, 0, -1, 35, 7]

# sorted() renvoie une NOUVELLE liste triée (la liste d'origine n'est pas modifiée)
print(sorted(test_list5))   # [-1, 0, 3, 4, 7, 35]
````

# Mission 6

Ask the user for a string, convert it to a list of characters with the `list` function, then ask for a single character.

Remove the first occurrence of the character from the list, then display how many occurrences of the character remain in the list.

For instance if you enter `"Excellent work"` then the letter `"e"`, it will display:
```
There are still 2 copies of e in the list
```

````python
texte = input("Entrez une chaîne : ")
caractere = input("Entrez un caractère : ")

# On met tout en minuscules : dans l'exemple, "Excellent work" + "e" donne 2 copies restantes,
# ce qui signifie que le "E" majuscule est compté comme un "e"
lettres = list(texte.lower())
caractere = caractere.lower()

lettres.remove(caractere)   # remove() supprime uniquement la PREMIÈRE occurrence
print(f"There are still {lettres.count(caractere)} copies of {caractere} in the list")
````

# Mission 7

Define three lists containing no duplicates. Then, indicate (however you like) the two that have the most elements in common (or all three if there's a tie).

For instance, with:
```
list_a = [1, 2, 4, 8, 16, 32]
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]
```
It may display:
```
Lists B and C have the most common elements!
```

````python
list_a = [1, 2, 4, 8, 16, 32]
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]

# Fonction qui compte les éléments communs à deux listes
def nb_communs(liste_1, liste_2):
    compteur = 0
    for element in liste_1:
        if element in liste_2:
            compteur += 1
    return compteur

ab = nb_communs(list_a, list_b)   # 3 : 1, 2, 8
ac = nb_communs(list_a, list_c)   # 1 : 2
bc = nb_communs(list_b, list_c)   # 4 : 2, 3, 5, 13
print("A-B :", ab, "| A-C :", ac, "| B-C :", bc)

maxi = max(ab, ac, bc)
if ab == ac == bc:
    print("Égalité : les trois listes ont autant d'éléments en commun !")
elif ab == maxi and ab > ac and ab > bc:
    print("Lists A and B have the most common elements!")
elif ac == maxi and ac > ab and ac > bc:
    print("Lists A and C have the most common elements!")
elif bc == maxi and bc > ab and bc > ac:
    print("Lists B and C have the most common elements!")
else:
    print("Égalité entre plusieurs paires : les trois listes sont à égalité !")
````
