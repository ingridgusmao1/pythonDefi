```python
%%writefile data_utils.py
"""
Module data_utils : fonctions mathématiques et statistiques de base.

Toutes les fonctions sont écrites "à la main" (sans le module math ni
statistics) pour s'entraîner à manipuler les boucles et les listes.
"""

# --- Constantes -------------------------------------------------------------
PI = 3.141592653589793   # nombre pi
E = 2.718281828459045    # nombre d'Euler (base du logarithme népérien)


# --- Fonctions de base ------------------------------------------------------
def carre(x):
    """Renvoie le carré de x (x * x)."""
    return x * x


def cube(x):
    """Renvoie le cube de x (x * x * x)."""
    return x * x * x


def factorielle(n):
    """
    Renvoie la factorielle de n (n! = 1 * 2 * ... * n).

    Paramètre : n (int >= 0)
    Retour    : int (par convention, 0! = 1)
    """
    if n < 0:
        raise ValueError("La factorielle n'est pas définie pour un nombre négatif.")
    resultat = 1
    for i in range(2, n + 1):
        resultat *= i
    return resultat


def valeur_absolue(x):
    """Renvoie la valeur absolue de x (x sans son signe)."""
    if x < 0:
        return -x
    return x


# --- Fonctions statistiques sur une liste -----------------------------------
def maximum(liste):
    """Renvoie le plus grand élément d'une liste non vide."""
    plus_grand = liste[0]
    for valeur in liste:
        if valeur > plus_grand:
            plus_grand = valeur
    return plus_grand


def minimum(liste):
    """Renvoie le plus petit élément d'une liste non vide."""
    plus_petit = liste[0]
    for valeur in liste:
        if valeur < plus_petit:
            plus_petit = valeur
    return plus_petit


def moyenne(liste):
    """Renvoie la moyenne arithmétique d'une liste non vide (somme / nombre d'éléments)."""
    total = 0
    for valeur in liste:
        total += valeur
    return total / len(liste)


def ecart_type(liste):
    """
    Renvoie l'écart-type (de population) d'une liste non vide :
    racine carrée de la moyenne des carrés des écarts à la moyenne.
    """
    m = moyenne(liste)
    somme_ecarts = 0
    for valeur in liste:
        somme_ecarts += (valeur - m) ** 2
    variance = somme_ecarts / len(liste)
    return variance ** 0.5   # puissance 0.5 = racine carrée


def mediane(liste):
    """
    Renvoie la médiane d'une liste non vide : la valeur centrale une fois
    la liste triée (ou la moyenne des deux valeurs centrales si la taille est paire).
    """
    triee = sorted(liste)          # sorted() ne modifie pas la liste d'origine
    n = len(triee)
    milieu = n // 2
    if n % 2 == 1:                 # nombre impair d'éléments
        return triee[milieu]
    return (triee[milieu - 1] + triee[milieu]) / 2
```

## Etape 1 : Refaites le process qu'on a vu ensemble dans les quêtes et au début du challenge pour créer une nouvelle version du module `data_utils_V2.py` en rajoutant de nouvelles fonctions.


*   Création du fichier `data_utils_V2.py`
*   Ajout et développement de nouvelles fonctions :
  * `somme_cumulee(liste)`: calcule la somme cumulée d'une liste et retourne une liste. Par exemple, pour la liste [1, 2, 3, 4], la somme cumulée serait [1, 3, 6, 10].

  * `compter_occurrences(liste)`: retourne un dictionnaire contenant le nombre d'occurrences de chaque élément de la liste. Par exemple, pour la liste [1, 2, 2, 3], le résultat serait {1: 1, 2: 2, 3: 1}.
  
  * `inverser_dict(dictionnaire)`: inverse les clés et les valeurs d'un dictionnaire. Par exemple, pour le dictionnaire {'a': 1, 'b': 2}, le résultat serait {1: 'a', 2: 'b'}.

  * `filtrer_pairs(liste)`: retourne une nouvelle liste contenant uniquement les nombres pairs de la liste d'entrée. Par exemple, pour la liste [1, 2, 3, 4], le résultat serait [2, 4].

  * `celsius_to_fahrenheit(celsius)`: convertit une température de Celsius en Fahrenheit et retourne un nombre. Par exemple, pour une température de 0°C, le résultat serait 32°F. *Rappel : la formule est F = (C * 9/5) + 32*

  * `trouver_doublons(liste)`: retourne une liste des éléments qui apparaissent plus d'une fois dans la liste d'entrée. Par exemple, pour la liste [1, 2, 2, 3, 3], le résultat serait [2, 3].

  * `calculer_moyenne_ponderee(valeurs, poids)`: calcule la moyenne pondérée d'une liste de valeurs avec une liste de poids correspondante et retourne un nombre. Par exemple, pour les valeurs [3, 5] et les poids [1, 2], le résultat serait 4.33.





```python
%%writefile data_utils_V2.py
"""
Module data_utils_V2 : nouvelle version du module data_utils.

Il reprend tout le contenu de data_utils (constantes et fonctions de base)
et ajoute de nouvelles fonctions de manipulation de listes et de dictionnaires.
"""

# On récupère tout ce qui existe déjà dans la V1 (PI, E, carre, cube, ...)
from data_utils import *


def somme_cumulee(liste):
    """
    Calcule la somme cumulée d'une liste de nombres.

    Paramètre : liste (list de nombres)
    Retour    : list, où chaque élément est la somme des éléments
                précédents + l'élément courant.
    Exemple   : [1, 2, 3, 4] -> [1, 3, 6, 10]
    """
    resultat = []
    total = 0
    for valeur in liste:
        total += valeur
        resultat.append(total)
    return resultat


def compter_occurrences(liste):
    """
    Compte le nombre d'apparitions de chaque élément d'une liste.

    Paramètre : liste (list)
    Retour    : dict {élément: nombre d'occurrences}
    Exemple   : [1, 2, 2, 3] -> {1: 1, 2: 2, 3: 1}
    """
    compteur = {}
    for element in liste:
        # get() renvoie 0 si la clé n'existe pas encore
        compteur[element] = compteur.get(element, 0) + 1
    return compteur


def inverser_dict(dictionnaire):
    """
    Inverse les clés et les valeurs d'un dictionnaire.

    Paramètre : dictionnaire (dict) dont les valeurs sont hachables et uniques
    Retour    : dict {valeur: clé}
    Exemple   : {'a': 1, 'b': 2} -> {1: 'a', 2: 'b'}
    Attention : si deux clés ont la même valeur, seule la dernière est conservée.
    """
    inverse = {}
    for cle, valeur in dictionnaire.items():
        inverse[valeur] = cle
    return inverse


def filtrer_pairs(liste):
    """
    Garde uniquement les nombres pairs d'une liste.

    Paramètre : liste (list d'entiers)
    Retour    : nouvelle list contenant les nombres pairs
    Exemple   : [1, 2, 3, 4] -> [2, 4]
    """
    pairs = []
    for nombre in liste:
        if nombre % 2 == 0:      # reste de la division par 2 nul => pair
            pairs.append(nombre)
    return pairs


def celsius_to_fahrenheit(celsius):
    """
    Convertit une température de degrés Celsius en degrés Fahrenheit.

    Paramètre : celsius (int ou float)
    Retour    : float, selon la formule F = C * 9/5 + 32
    Exemple   : 0 -> 32.0
    """
    return celsius * 9 / 5 + 32


def trouver_doublons(liste):
    """
    Trouve les éléments qui apparaissent plus d'une fois dans une liste.

    Paramètre : liste (list)
    Retour    : list des éléments en double (chacun une seule fois,
                dans l'ordre de leur première apparition)
    Exemple   : [1, 2, 2, 3, 3] -> [2, 3]
    """
    occurrences = compter_occurrences(liste)
    doublons = []
    for element, nombre in occurrences.items():
        if nombre > 1:
            doublons.append(element)
    return doublons


def calculer_moyenne_ponderee(valeurs, poids):
    """
    Calcule la moyenne pondérée d'une liste de valeurs.

    Paramètres : valeurs (list de nombres), poids (list de nombres, même longueur)
    Retour     : float = somme(valeur * poids) / somme(poids), arrondi à 2 décimales
    Exemple    : valeurs [3, 5], poids [1, 2] -> (3*1 + 5*2) / 3 = 4.33
    """
    if len(valeurs) != len(poids):
        raise ValueError("Les listes 'valeurs' et 'poids' doivent avoir la même longueur.")
    somme_ponderee = 0
    somme_poids = 0
    for v, p in zip(valeurs, poids):   # zip parcourt les deux listes en parallèle
        somme_ponderee += v * p
        somme_poids += p
    return round(somme_ponderee / somme_poids, 2)


# --- Partie bonus -----------------------------------------------------------
def analyser_texte(texte):
    """
    Analyse une chaîne de caractères.

    Paramètre : texte (str)
    Retour    : dict contenant :
        - 'nombre_mots'                : nombre total de mots
        - 'nombre_caracteres'          : nombre de caractères, espaces inclus
        - 'nombre_caracteres_sans_esp' : nombre de caractères, espaces exclus
        - 'mot_le_plus_long'           : le mot le plus long (sans ponctuation)
        - 'longueur_moyenne_mots'      : longueur moyenne des mots (2 décimales)
    """
    mots = texte.split()                                   # découpe sur les espaces
    # On retire la ponctuation collée aux mots (ex : "apprendre." -> "apprendre")
    mots_propres = [mot.strip(".,;:!?«»()\"") for mot in mots]

    mot_le_plus_long = ""
    total_lettres = 0
    for mot in mots_propres:
        total_lettres += len(mot)
        if len(mot) > len(mot_le_plus_long):
            mot_le_plus_long = mot

    return {
        "nombre_mots": len(mots),
        "nombre_caracteres": len(texte),
        "nombre_caracteres_sans_esp": len(texte.replace(" ", "")),
        "mot_le_plus_long": mot_le_plus_long,
        "longueur_moyenne_mots": round(total_lettres / len(mots), 2),
    }
```
