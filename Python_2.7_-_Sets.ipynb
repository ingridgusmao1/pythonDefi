It looks like a `dictionary`, but it's not a dictionary.

```python
my_set = {"apple", "kiwi", "pineapple"}
```

```python
type(my_set)
```

Question : Why isn't `my_set` an object from the `dict` class ?

Answer :
Parce qu'il ne contient que des **valeurs seules**, sans paires `clé: valeur`. Python décide du type selon le contenu des accolades :
- `{"apple", "kiwi"}` (éléments séparés par des virgules) → **`set`**
- `{"a": 1, "b": 2}` (paires avec `:`) → **`dict`**

Piège : `{}` vide crée un **dictionnaire** vide ; pour un set vide il faut écrire `set()`.

```python
print(type({"apple", "kiwi"}))   # set
print(type({"a": 1}))            # dict
print(type({}))                  # dict (et non set !)
print(type(set()))               # set
```

There are four collection data types in the Python programming language:

* `List` is a collection which is ordered and changeable. Allows duplicate members.
* `Tuple` is a collection which is ordered and unchangeable. Allows duplicate members.
* `Set` is a collection which is unordered, unchangeable, and unindexed. No duplicate members, but you can remove items and add new items.
* `Dictionary` is a collection which is ordered and changeable. No duplicate members.

Question :    

If you display a `set` object, it won't display with the same order you define it.

What can you say about the order the element are display ?

Answer :
L'ordre d'affichage **ne correspond pas à l'ordre de création et n'est pas garanti** : un set n'est pas ordonné. Python range les éléments selon leur **valeur de hachage (hash)**.
- Pour les petits entiers, le hash est égal au nombre lui-même, donc ils apparaissent souvent triés (1, 2, 3…), mais c'est un effet de bord.
- Pour les chaînes, le hash change à chaque nouvelle session Python : l'ordre peut donc varier d'une exécution à l'autre.

On ne doit donc **jamais compter sur l'ordre** des éléments d'un set (si on a besoin d'un ordre, on utilise `sorted(mon_set)`).

Question :   

Find a way to display the elements that exist in both sets.

```python
a_set = {"banana", "grapes", "cherry", "pineapple", "kiwi", "strawberry"}

b_set = {"lemon", "kiwi", "grapes", "blueberry", "watermelon"}

# Intersection : éléments présents dans les deux sets
print(a_set.intersection(b_set))
print(a_set & b_set)   # écriture équivalente avec l'opérateur &
```
