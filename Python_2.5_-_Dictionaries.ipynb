# Mission 1

Below is a dictionary containing the number of French campuses that offer each kind of Wild Code School course.

wildcodeschool = {
   "data_analyst": 4,
   "data_scientist": 2,
   "dev_web": "13"
}

Find a way to only display the keys that have int values.

```python
wildcodeschool =	{
  "data_analyst": 4,
  "data_scientist": 2,
  "dev_web": "13"
}
```

```python
# items() renvoie les paires (clé, valeur)
for cle, valeur in wildcodeschool.items():
    if type(valeur) == int:      # on garde uniquement les valeurs de type int
        print(cle)
```

For the same dictionary, find a way to display all the values that have a key that contains the letter "e".

```python
for cle, valeur in wildcodeschool.items():
    if "e" in cle:               # la clé contient-elle la lettre "e" ?
        print(valeur)
```

# Mission 3

Within the same dictionary, change the type of the value of the "dev_web" key to be an int. Then add a new key-value pair which will be "total_campus" : 19, using a **dictionary method**.

**Change the value of "dev_web" to be and int :**

```python
# On convertit la valeur "13" (str) en 13 (int) et on la réaffecte à la même clé
wildcodeschool["dev_web"] = int(wildcodeschool["dev_web"])
print(wildcodeschool)
```

**Add a new key-value pair which will be "total_campus" : 19, using a dictionary method**

```python
# update() est une méthode de dictionnaire qui ajoute (ou met à jour) des paires clé-valeur
wildcodeschool.update({"total_campus": 19})
print(wildcodeschool)
```

# Mission 4

Take the dictionary generated from mission 3. Delete the key value pair :
- 'dev_web': 13 using a method
- 'total_campus': 19 using another method

```python
# Méthode 1 : pop() supprime la clé indiquée (et renvoie sa valeur)
wildcodeschool.pop("dev_web")
print(wildcodeschool)

# Méthode 2 : popitem() supprime la DERNIÈRE paire ajoutée, ici "total_campus"
wildcodeschool.popitem()
print(wildcodeschool)
```

# Mission 6

Create a script that returns the same elements as the **values()** method, as a list.

Example: for `{'a': 1, 'b': 2, 'c': 3}` it will create `[1, 2, 3]`

```python
dico = {'a': 1, 'b': 2, 'c': 3}

# On reproduit values() : on parcourt les clés et on récupère chaque valeur
valeurs = []
for cle in dico:
    valeurs.append(dico[cle])

print(valeurs)
```
