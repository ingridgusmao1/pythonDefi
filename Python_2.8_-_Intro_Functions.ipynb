Create a function that returns the sum of 3 + 3 when executed.  
You must use the `return` statement.

```python
def somme_3_3():
    """Renvoie la somme de 3 + 3."""
    return 3 + 3

print(somme_3_3())   # 6
```

# Mission 2

Create a function that returns the sum of two numbers.  
The user of the function must be able to sum two numbers of his choice, when the function is executed.  
You must not use `input`.  


```python
def new_sum(a, b):
    """Renvoie la somme des deux nombres passés en arguments."""
    return a + b

print(new_sum(4, 7))      # 11
print(new_sum(2.5, 10))   # 12.5
```

Question:   
What is a parameter in a function?

Answer:
Un **paramètre** est une variable déclarée entre les parenthèses lors de la **définition** de la fonction (ici `a` et `b` dans `def new_sum(a, b)`). C'est un « emplacement » qui recevra une valeur.
La valeur concrète que l'on transmet lors de l'**appel** de la fonction s'appelle un **argument** (ici `4` et `7` dans `new_sum(4, 7)`). Grâce aux paramètres, la même fonction peut travailler avec des valeurs différentes à chaque appel.

# Mission 3

Create a function that returns the sum of two numbers chosen by the user of the function, in a sentence.  
For example:  
`new_sum_2(3,8)` returns `"The sum of 3 + 8 is equal to 11."`

```python
def new_sum_2(a, b):
    """Renvoie une phrase indiquant la somme de a et b."""
    return f"The sum of {a} + {b} is equal to {a + b}."

print(new_sum_2(3, 8))   # The sum of 3 + 8 is equal to 11.
```

# Mission 4

Create a function that returns 2 objects when executed:
* An object that will be the product of two numbers chosen by the user of the function.
* A word chosen by the user of the function.

```python
def produit_et_mot(a, b, mot):
    """Renvoie le produit de a et b, ainsi qu'un mot choisi par l'utilisateur."""
    return a * b, mot

resultat = produit_et_mot(6, 7, "Python")
print(resultat)         # (42, 'Python')
print(type(resultat))   # <class 'tuple'>
```

Your result must be two objects that are in brackets.
This is a tuple.

Question:
Why is the returned object a tuple that contains two objects?

Answer:
Lorsqu'on écrit plusieurs valeurs séparées par des **virgules** après `return` (`return a * b, mot`), Python les **regroupe automatiquement dans un tuple** : c'est la virgule qui crée le tuple, pas les parenthèses. Une fonction ne renvoie en réalité qu'**un seul objet** ; pour renvoyer plusieurs valeurs, elle les emballe donc dans un tuple. On peut ensuite les récupérer séparément : `produit, mot = produit_et_mot(6, 7, "Python")`.

# Mission 5

Create a function that returns the sum of two numbers when executed.  
Also, your function must return a sentence that simply indicates if the sum of the two numbers chosen by the user is:
* less than 10
* between 10 and 50
* greater than 50  
Again, you must use `return`.

```python
def somme_et_message(a, b):
    """Renvoie la somme de a et b et une phrase qui situe cette somme."""
    somme = a + b
    if somme < 10:
        message = "La somme est inférieure à 10."
    elif somme <= 50:
        message = "La somme est comprise entre 10 et 50."
    else:
        message = "La somme est supérieure à 50."
    return somme, message

print(somme_et_message(3, 4))     # (7, 'La somme est inférieure à 10.')
print(somme_et_message(20, 15))   # (35, 'La somme est comprise entre 10 et 50.')
print(somme_et_message(40, 30))   # (70, 'La somme est supérieure à 50.')
```
